```python
import pandas as pd
df = pd.read_csv("pilot.csv")
pattern = r"\bai\b|a\.i\.|ai-generated|generative|genai|slop|midjourney|chatgpt|artificial intelligence"
hits = df[df["review"].str.contains(pattern, case=False, regex=True)]
print(f"{len(hits)} of {len(df)} reviews mention AI ({len(hits)/len(df):.1%})")
print(hits.groupby("appid").size())
for text in hits["review"].head(10):
    print("-", text[:200])
```

```python
for i, row in hits.iterrows():
    print(f"[{i}] app {row['appid']}:\n{row['review']}\n")
```
